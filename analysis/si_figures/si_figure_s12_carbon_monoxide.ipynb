# SI Figure S12: MagNET vs DFT ¹³C shielding of carbon monoxide across bond lengths

Carbon monoxide stretched/compressed over a wide range of bond lengths, ¹³C shielding by DFT
(PBE1PBE/pcSseg-1, gas) vs the MagNET foundation model. The right panel shades the bond-length
distribution seen in real molecules.

```python
import os, sys

# make the in-repo modules importable (not pip-installed)
REPO = os.path.abspath("../..")
for _p in ("analysis/code", "analysis/code/shared"):
    sys.path.insert(0, os.path.join(REPO, _p))
```

```python
import matplotlib.pyplot as plt

import carbon_monoxide
```

```python
def figure_path(name):
    os.makedirs("figures", exist_ok=True)
    return os.path.join("figures", name)
```

```python
df = carbon_monoxide.load_comparison()
near, tails = carbon_monoxide.equilibrium_vs_extreme_error(df)
print(f'{len(df)} bond lengths from {df.bond_length.min():.2f} to {df.bond_length.max():.2f} angstrom')
print(f'mean |carbon error|: {near:.1f} ppm near equilibrium vs {tails:.1f} ppm in the tails')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ax = axes[0]
ax.plot(df["bond_length"], df["gaussian_c"], marker="o", linestyle="-", color="black", label="Gaussian C")
ax.plot(df["bond_length"], df["magnet_c"], marker="o", linestyle="-", color="gray", label="MagNET C")
ax.set_xlabel("Bond length (Å)", fontweight="bold")
ax.set_ylabel("Shielding (ppm)", fontweight="bold")
ax.set_title("13C Shielding vs. Bond Length", fontweight="bold")
ax.legend(frameon=False)

ax = axes[1]
centers, counts, bin_width = carbon_monoxide.load_bond_length_histogram()
ax_hist = ax.twinx()
ax_hist.bar(centers, counts, width=bin_width, color="0.75", edgecolor="none", zorder=0)
ax_hist.set_yticks([])
ax_hist.grid(False)
ax_hist.set_zorder(0)
ax.set_zorder(1)
ax.patch.set_alpha(0)   # let the histogram show through the transparent foreground axes
ax.plot(df["bond_length"], df["delta_c"], marker="o", linestyle="-", color="black")
ax.axhline(0, color="gray", linewidth=1, linestyle="--")
ax.set_xlabel("Bond length (Å)", fontweight="bold")
ax.set_ylabel("Δ shielding (MagNET − Gaussian, ppm)", fontweight="bold")
ax.set_title("Shielding Differences vs. Bond Lengths", fontweight="bold")

fig.tight_layout()
fig.savefig(figure_path("si_figure_s12.png"), dpi=300, bbox_inches="tight")
plt.show()
```
